```python
import torch
```

```python
tensor1 = torch.rand(7, 7)
tensor2 = torch.rand(1, 7)

tensor1, tensor2
```

```python
# tensor 2 change to Tranpose because matriks rand shape 1, 7 not compatible for 7 x 7
tensor2 = tensor2.T

tensor2
```

```python
tensorMulti = torch.matmul(tensor1, tensor2)

tensorMulti
```

```python
RANDOM_SEED = 0

torch.manual_seed(RANDOM_SEED)
tensor1 = torch.rand(7, 7)
tensor2 = torch.rand(1, 7)

tensor1, tensor2
```

```python
tensorMulti = torch.matmul(tensor1, tensor2.T)
tensorMulti
```

GPU Tensor

```python
device = "cuda"
RANDOM_SEED_GPU = 1234

torch.cuda.manual_seed(RANDOM_SEED_GPU)
tensor_gpu1 = torch.rand(2,3, device=device)
tensor_gpu2 = torch.rand(2,3, device=device)

tensor_gpu1, tensor_gpu2
print(tensor_gpu1.device)
```

```python
tensorMulti_GPU = torch.matmul(tensor_gpu1, tensor_gpu2.T)
tensorMulti_GPU
```

```python
print(f'Tensor Terkecil : {torch.min(tensorMulti_GPU)}')
print(f'Posisi Terkecil : {torch.argmin(tensorMulti_GPU)}')
print(f'Tensor Terbesar : {torch.max(tensorMulti_GPU)}')
print(f'Posisi Terbesar : {torch.argmax(tensorMulti_GPU)}')
```
